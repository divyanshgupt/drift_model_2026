```python
import numpy as np
from matplotlib import pyplot as plt

plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams['font.size'] = 12
plt.rcParams['lines.linewidth'] = 2
```

```python
## Plot two random walks with multiple seeds
# random walk 1 - small step size
# random walk 2 - large step size

random_walk_1 = np.cumsum(np.random.normal(loc=0, scale=0.1, size=(1000, 10)), axis=0)
random_walk_2 = np.cumsum(np.random.normal(loc=0, scale=0.2, size=(1000, 10)), axis=0)

plt.figure(figsize=(5, 4), dpi=300)
for i in range(random_walk_1.shape[1]):
    plt.plot(random_walk_1[:, i], alpha=0.5, color='blue')
plt.title('Random Walk 1 (Small Step Size)')
for i in range(random_walk_2.shape[1]):
    plt.plot(random_walk_2[:, i], alpha=0.5, color='orange')
plt.title('Random Walk 2 (Large Step Size)')
plt.tight_layout()
plt.show()
```
